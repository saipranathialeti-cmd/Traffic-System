```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import mutual_info_classif
```

```python
df = pd.read_csv("../data/traffic_dataset.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
#Select Features and Target
features = [
    "date",
    "timestamp",
    "direction",
    "day_night",
    "weather",
    "start_frame",
    "number_of_frames"
]

X = df[features].copy()
y = df["traffic_class"]

print("Selected features:")
print(features)
print("\nTarget:", "traffic_class")
```

```python
#Create New Feature
X["end_frame"] = X["start_frame"] + X["number_of_frames"]

print("New feature created: end_frame")
X.head()
```

```python
#Encode Categorical Features
X = pd.get_dummies(
    X,
    columns=["direction", "day_night", "weather"],
    drop_first=False
)

X = X.astype(int)

print("Categorical features encoded successfully!")
print(X.head())
```

```python
#Encoder Target
encoder = LabelEncoder()

y_encoded = encoder.fit_transform(y)

print("Classes:", encoder.classes_)
print("Encoded classes:", sorted(set(y_encoded)))
```

```python
#Feature Importance
mi = mutual_info_classif(X, y_encoded, random_state=42)

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Score": mi
}).sort_values("Score", ascending=False)

print(feature_importance)
```

```python
#Final Features
final_features = X.columns.tolist()

print("FINAL FEATURES:")
for feature in final_features:
    print("-", feature)

print("\nTotal features:", len(final_features))
```

## Final observation
Important features were selected for building the traffic classification model.
Categorical features such as direction, day/night, and weather were converted into numerical form.
The target variable traffic_class was used for classification.
Feature importance was analyzed to understand which features contribute to prediction.
The final processed features were prepared for model training.
